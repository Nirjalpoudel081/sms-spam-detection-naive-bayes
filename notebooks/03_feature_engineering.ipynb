```python

import pandas as pd
from pathlib import Path

from sklearn.feature_extraction.text import TfidfVectorizer

# Paths to our saved training and testing data
DATA_DIR = Path("../data/processed")

# Load the datasets created in Notebook 02
train_df = pd.read_csv(DATA_DIR / "train.csv")
test_df = pd.read_csv(DATA_DIR / "test.csv")

# Inspect the loaded datasets
print("Training data shape:", train_df.shape)
print("Testing data shape:", test_df.shape)

train_df.head()
```

```python

# Separate message text from target labels

X_train = train_df["message"]
y_train = train_df["label"]

X_test = test_df["message"]
y_test = test_df["label"]

# Verify the data
print("Training messages:", X_train.shape)
print("Training labels:", y_train.shape)

print("Testing messages:", X_test.shape)
print("Testing labels:", y_test.shape)
```

```python

# Initialize the TF-IDF vectorizer
tfidf_vectorizer = TfidfVectorizer()

# Learn the vocabulary from training messages and transform them
X_train_tfidf = tfidf_vectorizer.fit_transform(X_train)

# Transform test messages using the same learned vocabulary
X_test_tfidf = tfidf_vectorizer.transform(X_test)

# Inspect the resulting feature matrices
print("Training TF-IDF shape:", X_train_tfidf.shape)
print("Testing TF-IDF shape:", X_test_tfidf.shape)

print("Vocabulary size:", len(tfidf_vectorizer.vocabulary_))
```

```python

import joblib

# Create a folder for saved model artifacts
MODEL_DIR = Path("../models")
MODEL_DIR.mkdir(parents=True, exist_ok=True)

# Save the fitted TF-IDF vectorizer
vectorizer_path = MODEL_DIR / "tfidf_vectorizer.joblib"

joblib.dump(tfidf_vectorizer, vectorizer_path)

print("TF-IDF vectorizer saved to:", vectorizer_path)
```

```python

from scipy.sparse import save_npz

# Save the TF-IDF matrices
save_npz(MODEL_DIR / "X_train_tfidf.npz", X_train_tfidf)
save_npz(MODEL_DIR / "X_test_tfidf.npz", X_test_tfidf)

# Save the labels
y_train.to_csv(MODEL_DIR / "y_train.csv", index=False)
y_test.to_csv(MODEL_DIR / "y_test.csv", index=False)

print("TF-IDF matrices and labels saved successfully.")
```
